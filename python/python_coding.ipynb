```python
df["Customer_Age"] = df["Customer_Age"].fillna(
    df["Customer_Age"].median()
)
```

```python
df["Discount"] = df["Discount"].fillna(
    df["Discount"].median()
)
```

```python
df["Payment_Method"] = df["Payment_Method"].fillna(
    df["Payment_Method"].mode()[0]
)
```

```python
Q1 = df["Revenue"].quantile(0.25)
Q3 = df["Revenue"].quantile(0.75)

IQR = Q3 - Q1

print("Q1:", Q1)
print("Q3:", Q3)
print("IQR:", IQR)
```

```python
lower_limit = Q1 - 1.5 * IQR
upper_limit = Q3 + 1.5 * IQR

print("Lower Limit:", lower_limit)
print("Upper Limit:", upper_limit)
```

```python
outliers = df[
    (df["Revenue"] < lower_limit) |
    (df["Revenue"] > upper_limit)
]

print("Number of outliers:", len(outliers))
```

```python
outliers.head(10)
```

```python
plt.figure(figsize=(8, 5))
sns.boxplot(x=df["Revenue"])
plt.title("Revenue Outlier Detection")
plt.show()
```

```python
outliers['Category'].value_counts()
```

```python
df["Calculated_Revenue"] = (
    df["Quantity"] *
    df["Unit_Price"] *
    (1 - df["Discount"])
).round(2)
```

```python
(df["Revenue"] == df["Calculated_Revenue"]).value_counts()
```

```python
df[df["Revenue"] != df["Calculated_Revenue"]][[
    "Order_ID",
    "Quantity",
    "Unit_Price",
    "Discount",
    "Revenue",
    "Calculated_Revenue"
]].head(20)
```

```python
mismatch = df["Revenue"] != df["Calculated_Revenue"]

print("Mismatched records:", mismatch.sum())
```

```python
df.loc[mismatch, "Revenue"] = df.loc[mismatch, "Calculated_Revenue"]
```

```python
(df["Revenue"] == df["Calculated_Revenue"]).value_counts()
```
